```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
dice = [1,2,3,4,5,6]

```

```python
lst = range(1, 100000, 10)

means = [np.random.choice(dice, replace = True, size = i).mean() 
         for i in lst]


_ = plt.plot(lst, means)
```

```python
x = np.random.choice(dice, replace = True, size = (1000, 10000))
y = np.random.choice(dice, replace = True, size = (1000, 1000))
```

```python
# _ = plt.hist(x.mean(axis = 1), alpha = 0.5)
_ = plt.hist(y.mean(axis = 1), alpha = 0.5)
```

# Нормальное распределение

```python
x = np.random.exponential(100, size = 1000)
_ = plt.hist(x, alpha = 0.5)
```

```python
x = np.random.exponential(100, size = (1000, 1000)).mean(axis = 1)
```

```python
_ = plt.hist(x, alpha = 0.5)
```

```python
x = np.random.poisson(0.5, size = 1000)
_ = plt.hist(x, alpha = 0.5)
```

```python
x = np.random.poisson(0.5, size = (1000, 1000))
```

```python
_ = plt.hist(x.mean(axis = 1), alpha = 0.5)
```

```python
x = np.random.exponential(100, size = 1000)
_ = plt.hist(x, alpha = 0.5)
```

```python
_ = plt.hist(np.hstack([np.random.exponential(100, size = (1000,1000)), 
                        np.random.exponential(100, size = (1000,1000))+400]).mean(axis = 1))
```

# mean vs quantile

```python
x = list(np.random.randint(1, 100000, size = 1000))
```

```python
y = x[:-1]+[10000000000]
```

```python
np.mean(x)
```

```python
np.mean(y)
```

```python
np.median(x)
```

```python
np.median(y)
```
